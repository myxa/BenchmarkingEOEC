```python
from pathlib import Path

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# Strategy Comparison

```python
def get_strategies_comparison(pth, site, atlas, fc_type, *, savepath=None, show=True):
    pth = Path(pth)
    filepath = pth / site / atlas / f"{site}_{atlas}_{fc_type}_strategies_comparison.csv"
    strat_comparison = pd.read_csv(filepath, index_col=0)

    fig, ax = plt.subplots(figsize=(20, 20))
    sns.heatmap(
        strat_comparison,
        annot=True,
        fmt=".2f",
        cmap="Blues",
        vmin=0,
        vmax=1,
        square=True,
        cbar=False,
        ax=ax,
    )

    fc_name = {"corr": "Correlation", "partial": "Partial correlation", 
               "tangent": "Tangent space parametrization", "glasso": "Graphical lasso"}
    
    ax.set_title(f"Strategies similarity for {site.upper()}, {atlas}, {fc_name[fc_type]}", fontsize=20)
    ax.set_xticklabels(ax.get_xticklabels(), rotation=40, ha="right")
    fig.tight_layout()

    if savepath is not None:
        savepath = Path(savepath)
        savepath = savepath / f"{site}_{atlas}_{fc_type}_strategies_comparison.png"
        savepath.parent.mkdir(parents=True, exist_ok=True)
        fig.savefig(savepath, dpi=300, bbox_inches="tight")

    if show:
        plt.show()
    else:
        plt.close(fig)

    return fig, ax
```

```python
for atlas in ['AAL', 'Schaefer200', 'Brainnetome']:
    for fc_type in ['corr', 'partial']:
        _ = get_strategies_comparison('/home/tm/benchmarking_data/results', 
                                        'china', atlas, fc_type, 
                                        savepath='/home/tm/benchmarking_data/results/figures', 
                                        show=False)
```
